```python
from bayesien_paths import BayesMap
from private_path_query_utils import Path
from typing import List
import random


def safe_path(grid : List[List[int]] , path : Path) -> bool:
    if grid[path.start[0]][path.start[1]] == 1:
        return False
    cx, cy = path.start
    for dx, dy in path.moves:
        cx += dx
        cy += dy
        if cx < 0 or cx >= len(grid):
            return False
        if grid[cx][cy] == 1:
            return False
    return True

grid_size = 10
# make a random grid of 1's and 0's with probability 0.2 of being 1
grid = [[1 if random.random() < 0.1 else 0 for _ in range(10)] for _ in range(10)]
grid[0][0] = 0
grid[grid_size-1][grid_size-1] = 0

print("ground truth grid:")
for row in grid:
    print(" ".join(str(cell) for cell in row))
    

print("Bayesien paths module loaded successfully.")
map = BayesMap(size=grid_size, p_init = 0.5)
print("BayesMap instance created successfully with size 10.")
print(map)


for i in range(100):
    print(f"\nIteration {i+1}:")
    # randomly find a start 

    result = map.find_highest_entropy_path(length = 8)
    print(result)
    path, start = result
    print(f"chosen start={start}")
    

    print("highest entropy path found:")
    print(path.pretty_str())
    print("checking safety of path...")
    path_safe = safe_path(grid, path)
    print("path is safe:", path_safe)

    print("updating map probs")
    
    map.update_probabilities(path = path, safe = path_safe)

    print("updated map:")
    print(map)
    print()
    print(f"Does there exist a path from (0,0) to ({grid_size-1},{grid_size-1})?")
    print(map.check_viable_path())



```
