````python
from logic4e import HybridWumpusAgent, WumpusKB, WumpusSATKB, percept_glitter
from wumpus_mat_sat import WumpusKBMatSat
from agents import Glitter, Bump, Stench, Breeze, Scream
from agents4e import WumpusEnvironment, WumpusTestEnvironment

LABELS = ["Stench", "Breeze", "Glitter", "Bump", "Scream"]
EVENT_CLASSES = [Stench, Breeze, Glitter, Bump, Scream]
print(percept_glitter(3))
def to_event(percept_tuple):
    if percept_tuple is None:
        return None
    flags = [bool(x) for x in percept_tuple]
    if not any(flags):
        return None
    if sum(flags) > 1:
        # Original KB only handles one positive percept at a time
        raise ValueError("Multiple positives in percept; logic4e expects one.")
    idx = flags.index(True)
    return EVENT_CLASSES[idx]()  # e.g., Breeze()

def show(step, percept, action):
    active = [lab for lab, v in zip(LABELS, percept) if v]
    name = ", ".join(active) if active else "None"
    print(f"t={step:02d}  percept={name:<20} -> action={action}")

agent = HybridWumpusAgent(dimentions=4, kb_class=WumpusKBMatSat)
env = WumpusTestEnvironment(agent.program, width=4, height=4, show=True)
env.run(4) 
````

## Multi-Agent System Architecture

The multi-agent test above demonstrates a collaborative Wumpus World system with the following components:

### Knowledge Base Architecture
- **Public KB**: Shared world rules and common knowledge (e.g., Wumpus death)
- **Private KBs**: Each agent stores their own percepts separately
- **Current Mode**: Agents query ALL private KBs (collaborative reasoning)

### Agent Components
1. **Logic-based reasoning**: Uses MatSat SAT solver for fast entailment checking
2. **Path planning**: A* search for optimal navigation through safe squares
3. **Decision hierarchy**: Gold grabbing → Exploration → Wumpus shooting → Exit

### Information Flow
```
Environment → (agent_id, percepts) → Agent Program
                                          ↓
                            Store in private_KB[agent_id]
                                          ↓
                            Query public_KB + all private_KBs
                                          ↓
                            Plan actions using A* search
                                          ↓
                            Return action to environment
```

### Key Features
- ✅ Each agent maintains separate percept history
- ✅ Agents can access shared knowledge for collaboration
- ✅ MatSat C++ solver for 10-100x faster inference
- ✅ Detailed debugging output showing world state and reasoning
